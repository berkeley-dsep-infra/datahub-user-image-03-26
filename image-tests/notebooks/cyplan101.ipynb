## highlight-text

```python
import matplotlib.pyplot as plt
from highlight_text import fig_text

fig, ax = plt.subplots(figsize=(6, 2))
ax.plot([0, 1, 2], [1, 3, 2])
ax.set_xticks([]); ax.set_yticks([])
fig_text(
    x=0.5, y=0.95,
    s="Hello <world>",
    highlight_textprops=[{"color": "red", "fontweight": "bold"}],
    ha="center", fontsize=14, fig=fig,
)
plt.show()
```

## contextily

```python
import contextily as cx
import geopandas as gpd
import matplotlib.pyplot as plt
from shapely.geometry import Point

gdf = gpd.GeoDataFrame(
    geometry=[Point(-122.2544, 37.8703)],
    crs="EPSG:4326",
).to_crs(epsg=3857)

ax = gdf.plot(figsize=(6, 6), color="red", markersize=80)
cx.add_basemap(ax, source=cx.providers.CartoDB.Positron)
ax.set_axis_off()
plt.show()
```

## lonboard

```python
import geopandas as gpd
from shapely.geometry import Point
from lonboard import Map, ScatterplotLayer

gdf = gpd.GeoDataFrame(
    geometry=[
        Point(-122.2544, 37.8703),
        Point(-122.2593, 37.8722),
    ],
    crs="EPSG:4326",
)
layer = ScatterplotLayer.from_geopandas(gdf, get_radius=25, get_fill_color=[200, 60, 60])
Map(layer)
```

## overturemaps

```python
from overturemaps import core

bbox = (-122.2730, 37.8680, -122.2680, 37.8720)
# Overture's STAC catalog (used to auto-resolve the latest release) has been
# intermittently unavailable, so pin a known-good release explicitly instead
# of relying on that lookup. Bump this periodically as Overture cuts new
# monthly releases: https://github.com/OvertureMaps/overturemaps-py
gdf = core.geodataframe("building", bbox=bbox, release="2026-07-22.0")
print(len(gdf), "buildings")
gdf.plot(figsize=(5, 5), edgecolor="white", linewidth=0.3)
```
